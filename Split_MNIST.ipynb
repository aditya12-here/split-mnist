```python
from sklearn.datasets import fetch_openml
```

```python
mnist=fetch_openml("mnist_784",as_frame=False)
```

```python
X,y=mnist.data,mnist.target
```

```python
import pandas as pd
import numpy as np
```

```python
y=y.astype(np.int64)
```

```python
# i will be creating tasks by myself
# for this i will take classes belonging to 0 and 1 and their corresponding images as task 1
# build another dataset for task 2 - 2,3
```

```python
combined=np.c_[X,y]
```

```python
task1=combined[np.isin(combined[:,-1],[0,1])]
```

```python
X_task1=task1[:,:-1]
y_task1=task1[:,-1]
```

```python
from sklearn.linear_model import SGDClassifier
```

```python
model=SGDClassifier(random_state=42)
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_task1_train,X_task1_test,y_task1_train,y_task1_test=train_test_split(X_task1,y_task1,test_size=0.2,random_state=42,stratify=y_task1)
```

```python
model.partial_fit(X_task1_train,y_task1_train,classes=np.arange(10))
```

```python
pred_task1=model.predict(X_task1_test)
```

```python
accuracy_over_tasks=[]
```

```python
from sklearn.metrics import accuracy_score
```

```python
accuracy_over_tasks.append(accuracy_score(pred_task1,y_task1_test))
```

```python
task2=combined[np.isin(combined[:,-1],[2,3])]
```

```python
X_task2=task2[:,:-1]
y_task2=task2[:,-1]
```

```python
X_task2_train,X_task2_test,y_task2_train,y_task2_test=train_test_split(X_task2,y_task2,test_size=0.2,random_state=42,stratify=y_task2)
```

```python
model.partial_fit(X_task2_train,y_task2_train)
```

```python
pred_task2=model.predict(X_task2_test)
```

```python
accuracy_over_tasks.append(accuracy_score(pred_task2,y_task2_test))
```

```python
mod=SGDClassifier(random_state=42)
```

```python
mod.fit(X_task2_train,y_task2_train)
```

```python
pred_task2_without_partial=mod.predict(X_task2_test)
```

```python
accuracy_score(pred_task2_without_partial,y_task2_test)
#2,3 hard to learn + this had no training on 0,1 still less accuracy could be just random_state
```
